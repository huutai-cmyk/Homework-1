```python
#2.11
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import numpy as np
Ride_Distance = ctrl.Antecedent(np.arange(0,50.1,0.1),'Ride Distance')
Traffic_Condition= ctrl.Antecedent(np.arange(0,100.1,0.1),'Traffic Condition')
Demand_level= ctrl.Antecedent(np.arange(0,100.1,0.1),'Demand level')
Weather_condition= ctrl.Antecedent(np.arange(0,10.1,0.1),'Weather condition')
Customer_rating=ctrl.Antecedent(np.arange(1,5.1,0.1),'Customer rating')
Ride_Punctuality=ctrl.Antecedent(np.arange(0,100.1,0.1),'Ride punctuality')
Price=ctrl.Consequent(np.arange(0,10.1,0.1),'Price')
Point=ctrl.Consequent(np.arange(0,10.1,0.1),'Point')

Ride_Distance['Ngắn'] = fuzz.trimf(Ride_Distance.universe,[0,0,3])
Ride_Distance['Trung bình'] = fuzz.trimf(Ride_Distance.universe,[2,4,8])
Ride_Distance['Dài'] = fuzz.trimf(Ride_Distance.universe,[6,12,20])
Ride_Distance['Rất dài'] = fuzz.trimf(Ride_Distance.universe,[15,50,50])
Traffic_Condition['Thấp'] = fuzz.trimf(Traffic_Condition.universe,[0,0,30])
Traffic_Condition[' Trung bình'] = fuzz.trimf(Traffic_Condition.universe,[20,40,60])
Traffic_Condition['Cao'] = fuzz.trimf(Traffic_Condition.universe,[60,100,100])
Demand_level['Thấp'] = fuzz.trimf(Demand_level.universe,[0,0,30])
Demand_level['Trung bình'] = fuzz.trimf(Demand_level.universe,[20,40,60])
Demand_level['Cao'] = fuzz.trimf(Demand_level.universe,[60,100,100])
Weather_condition['Good'] = fuzz.trimf(Weather_condition.universe,[0,0,5])
Weather_condition['Normal'] = fuzz.trimf(Weather_condition.universe,[0,5,10])
Weather_condition['Bad'] = fuzz.trimf(Weather_condition.universe,[5,10,10])
Customer_rating['Kém'] = fuzz.trimf(Customer_rating.universe,[1,1,2.5])
Customer_rating['Trung bình'] = fuzz.trimf(Customer_rating.universe,[2,3,4])
Customer_rating['Tốt'] = fuzz.trimf(Customer_rating.universe,[3.5,5,5])
Ride_Punctuality['Trễ'] = fuzz.trimf(Ride_Punctuality.universe,[0,0,50])
Ride_Punctuality['Đúng Giờ'] = fuzz.trimf(Ride_Punctuality.universe,[40,80,80])
Ride_Punctuality['Sớm'] = fuzz.trimf(Ride_Punctuality.universe,[70,100,100])
Ride_Distance.view()
Price['thấp']=fuzz.trimf(Price.universe,[0,0,3])
Price['Trung bình']=fuzz.trimf(Price.universe,[0,2,6])
Price['Cao']=fuzz.trimf(Price.universe,[5,8,10])
Price['Rất cao']=fuzz.trimf(Price.universe,[8,10,10])
Point['Không có']=fuzz.trimf(Point.universe,[0,0,0])
Point['Ít']=fuzz.trimf(Point.universe,[0,0,5])
Point['Trung bình']=fuzz.trimf(Point.universe,[0,5,10])
Point['Cao']=fuzz.trimf(Point.universe,[5,10,10])
Price.view()
Point.view()
Traffic_Condition.view()
Demand_level.view()
Weather_condition.view()
Customer_rating.view()
Ride_Punctuality.view()
```

```python
rule1= ctrl.Rule(Ride_Distance['Ngắn'] & Traffic_Condition['Thấp'] & Demand_level['Thấp'], Price['thấp'] )
rule2= ctrl.Rule(Ride_Distance['Ngắn'] & Traffic_Condition[' Trung bình'] & Demand_level['Cao'], Price['Trung bình'] )
rule3= ctrl.Rule(Ride_Distance['Trung bình'] & Traffic_Condition['Cao'] & Demand_level['Cao'], Price['Cao'] )
rule4= ctrl.Rule(Ride_Distance['Dài'] & Traffic_Condition[' Trung bình'] & Weather_condition['Good'], Price['Trung bình'] )
rule5= ctrl.Rule(Ride_Distance['Dài'] & Traffic_Condition['Cao'] & Weather_condition['Bad'], Price['Rất cao'] )
rule6= ctrl.Rule(Ride_Distance['Rất dài'] & Traffic_Condition['Cao'] & Demand_level['Cao'],Price['Cao'] )
rule7= ctrl.Rule(Ride_Distance['Trung bình'] & Traffic_Condition['Thấp'] & Demand_level['Thấp'],Price['Trung bình'] )
rule8= ctrl.Rule(Ride_Distance['Ngắn'] & Traffic_Condition['Cao'] & Weather_condition['Bad'],Price['Cao'])
rule9= ctrl.Rule(Ride_Distance['Rất dài'] & Weather_condition['Bad'],Price['Rất cao'])
rule10= ctrl.Rule(Ride_Distance['Trung bình'] & Traffic_Condition[' Trung bình'] & Weather_condition['Normal'],Price['Trung bình'])
rule11 = ctrl.Rule(Customer_rating['Tốt'] & Ride_Punctuality['Đúng Giờ'], Point['Cao'])
rule12 = ctrl.Rule(Customer_rating['Trung bình'] & Ride_Punctuality['Đúng Giờ'], Point['Trung bình'])
rule13 = ctrl.Rule(Customer_rating['Kém'] & Ride_Punctuality['Trễ'], Point['Không có'])
rule14=ctrl.Rule(Ride_Distance['Dài'] & Traffic_Condition['Cao'] & Ride_Punctuality['Đúng Giờ'],Price['Cao'])
rule15=ctrl.Rule(Ride_Distance['Rất dài'] & Traffic_Condition['Cao'] & Ride_Punctuality['Trễ'],Price['Rất cao'])
rule16=ctrl.Rule(Customer_rating['Kém'] & Ride_Punctuality['Trễ'],Point['Không có'])
rule17=ctrl.Rule(Ride_Distance['Rất dài'] & Weather_condition['Bad'] & Customer_rating['Tốt'], Point['Cao'] )
rule18=ctrl.Rule(Ride_Distance['Ngắn'] & Customer_rating['Trung bình'] & Ride_Punctuality['Đúng Giờ'], Point['Ít'] )
rule19=ctrl.Rule(Ride_Distance['Dài'] & Traffic_Condition['Cao'] & Customer_rating['Trung bình'] & Ride_Punctuality['Đúng Giờ'], Point['Ít'] )
rule20=ctrl.Rule(Ride_Distance['Trung bình'] & Weather_condition['Normal'] & Customer_rating['Tốt'] , Point['Trung bình'] )
Calculate_ctrl=ctrl.ControlSystem([
    rule1,rule2,rule3,rule4,rule5,rule6,rule7,rule8,rule9,rule10,rule11,rule12,rule13,rule14,rule15,rule16,
    rule17,rule18,rule19,rule20
])
Calculate=ctrl.ControlSystemSimulation(Calculate_ctrl)

Calculate.input['Ride Distance']=15
Calculate.input['Traffic Condition']=80
Calculate.input['Demand level']=80
Calculate.input['Weather condition']=6
Calculate.input['Customer rating']=4
Calculate.input['Ride punctuality']=80
Calculate.compute()
print(Calculate.output)
Price.view(sim=Calculate)
print(Calculate.output['Price'])
```

```python
#2.12
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import numpy as np
Đánh_Giá=ctrl.Antecedent(np.arange(1,5.1,0.1),'Đánh Giá')
Khối_Lượng_Bán=ctrl.Antecedent(np.arange(0,10.1,0.1),'Khối Lượng Bán')
Biên_Lợi_Nhuận=ctrl.Antecedent(np.arange(0,10.1,0.1),'Biên Lợi Nhuận')
Sự_Kiện_Theo_Mùa=ctrl.Antecedent(np.arange(0,10.1,0.1),'Sự Kiện Theo Mùa')
Competitor_Discounts=ctrl.Antecedent(np.arange(0,10.1,0.1),'Competitor Discounts')
Rate=ctrl.Consequent(np.arange(0,70.1,0.1),'Rate')

Đánh_Giá['Thấp']=fuzz.trimf(Đánh_Giá.universe,[0,0,4])
Đánh_Giá['Trung Bình']=fuzz.trimf(Đánh_Giá.universe,[4,4.5,4.5])
Đánh_Giá['Cao']=fuzz.trimf(Đánh_Giá.universe,[4.5,5,5])
Khối_Lượng_Bán['Thấp']=fuzz.trimf(Khối_Lượng_Bán.universe,[0,0,5])
Khối_Lượng_Bán['Trung Bình']=fuzz.trimf(Khối_Lượng_Bán.universe,[0,5,10])
Khối_Lượng_Bán['Cao']=fuzz.trimf(Khối_Lượng_Bán.universe,[5,10,10])
Biên_Lợi_Nhuận['Thấp']=fuzz.trimf(Biên_Lợi_Nhuận.universe,[0,0,5])
Biên_Lợi_Nhuận['Trung Bình']=fuzz.trimf(Biên_Lợi_Nhuận.universe,[0,5,10])
Biên_Lợi_Nhuận['Cao']=fuzz.trimf(Biên_Lợi_Nhuận.universe,[5,10,10])
Sự_Kiện_Theo_Mùa['Không']=fuzz.trimf(Sự_Kiện_Theo_Mùa.universe,[0,0,0])
Sự_Kiện_Theo_Mùa['Trung Bình']=fuzz.trimf(Sự_Kiện_Theo_Mùa.universe,[0,0,5])
Sự_Kiện_Theo_Mùa['Cao']=fuzz.trimf(Sự_Kiện_Theo_Mùa.universe,[5,10,10])
Competitor_Discounts['Ít']=fuzz.trimf(Competitor_Discounts.universe,[0,0,3])
Competitor_Discounts['Trung Bình']=fuzz.trimf(Competitor_Discounts.universe,[0,3,6])
Competitor_Discounts['Cao']=fuzz.trimf(Competitor_Discounts.universe,[6,10,10])
Rate['Rất Thấp']=fuzz.trimf(Rate.universe,[0,0,5])
Rate['Thấp']=fuzz.trimf(Rate.universe,[5,10,10])
Rate['Trung Bình']=fuzz.trimf(Rate.universe,[10,20,20])
Rate['Cao']=fuzz.trimf(Rate.universe,[20,40,40])
Rate['Rất Cao']=fuzz.trimf(Rate.universe,[40,70,70])

Đánh_Giá.view()
Khối_Lượng_Bán.view()
Biên_Lợi_Nhuận.view()
Sự_Kiện_Theo_Mùa.view()
Competitor_Discounts.view()
Rate.view()


```

```python
rule1=ctrl.Rule(Đánh_Giá['Cao'] & Khối_Lượng_Bán['Cao'] & Biên_Lợi_Nhuận['Cao'], Rate['Rất Thấp'])
rule2=ctrl.Rule(Đánh_Giá['Thấp'] & Khối_Lượng_Bán['Thấp'] & Biên_Lợi_Nhuận['Cao'], Rate['Cao'])
rule3=ctrl.Rule(Sự_Kiện_Theo_Mùa['Cao'] & Competitor_Discounts['Cao'], Rate['Rất Cao'])
rule4=ctrl.Rule(Đánh_Giá['Trung Bình'] & Khối_Lượng_Bán['Trung Bình'] & Biên_Lợi_Nhuận['Trung Bình'], Rate['Trung Bình'])
rule5=ctrl.Rule(Competitor_Discounts['Ít'] & Khối_Lượng_Bán['Cao'] & Biên_Lợi_Nhuận['Thấp'], Rate['Trung Bình'])
rule6=ctrl.Rule(Đánh_Giá['Thấp'] & Sự_Kiện_Theo_Mùa['Không'], Rate['Trung Bình'])
rule7=ctrl.Rule(Khối_Lượng_Bán['Thấp'] & Biên_Lợi_Nhuận['Thấp'], Rate['Rất Cao'])

Chiết_khấu_ctrl=ctrl.ControlSystem([
    rule1,rule2,rule3,rule4,rule5,rule6,rule7
])
Chiết_khau=ctrl.ControlSystemSimulation(Chiết_khấu_ctrl)
Chiết_khau.input['Đánh Giá']=5
Chiết_khau.input['Khối Lượng Bán']=10
Chiết_khau.input['Biên Lợi Nhuận']=10
Chiết_khau.input['Sự Kiện Theo Mùa']=10
Chiết_khau.input['Competitor Discounts']=10
Chiết_khau.compute()
print(Chiết_khau.output)
Rate.view(sim=Chiết_khau)
print(Chiết_khau.output['Rate'])
```

```python
#2.14
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import numpy as np
Mật_Độ_Đơn=ctrl.Antecedent(np.arange(0,10.1,0.1),'Mật độ đơn')
Delivery_Urgency=ctrl.Antecedent(np.arange(0,10.1,0.1),'Delivery_Urgency')
Tải_Trọng=ctrl.Antecedent(np.arange(0,10.1,0.1),'Tải trọng')
Traffic_Condition=ctrl.Antecedent(np.arange(0,10.1,0.1),'Traffic_Condition')
Demand_level=ctrl.Antecedent(np.arange(0,10.1,0.1),'Demand_level')
Lợi_Nhuận=ctrl.Antecedent(np.arange(0,10.1,0.1),'Lợi nhuận')
Đơn_kết_Hợp=ctrl.Consequent(np.arange(0,10.1,0.1),'Đơn kết hợp')
Ưu_Tiên=ctrl.Consequent(np.arange(0,10.1,0.1),'Ưu Tiên')

Mật_Độ_Đơn['Thấp']=fuzz.trimf(Mật_Độ_Đơn.universe,[0,0,3])
Mật_Độ_Đơn['Trung Bình']=fuzz.trimf(Mật_Độ_Đơn.universe,[2,4,7])
Mật_Độ_Đơn['Cao']=fuzz.trimf(Mật_Độ_Đơn.universe,[8,10,10])
Delivery_Urgency['Thấp']=fuzz.trimf(Delivery_Urgency.universe,[0,0,3])
Delivery_Urgency['Trung Bình']=fuzz.trimf(Delivery_Urgency.universe,[2,4,7])
Delivery_Urgency['Cao']=fuzz.trimf(Delivery_Urgency.universe,[8,10,10])
Tải_Trọng['Thấp']=fuzz.trimf(Tải_Trọng.universe,[0,0,3])
Tải_Trọng['Trung Bình']=fuzz.trimf(Tải_Trọng.universe,[2,4,7])
Tải_Trọng['Cao']=fuzz.trimf(Tải_Trọng.universe,[8,10,10])
Traffic_Condition['Thấp']=fuzz.trimf(Traffic_Condition.universe,[0,0,3])
Traffic_Condition['Trung Bình']=fuzz.trimf(Traffic_Condition.universe,[2,4,7])
Traffic_Condition['Cao']=fuzz.trimf(Traffic_Condition.universe,[8,10,10])
Lợi_Nhuận['Thấp']=fuzz.trimf(Lợi_Nhuận.universe,[0,0,3])
Lợi_Nhuận['Trung Bình']=fuzz.trimf(Lợi_Nhuận.universe,[2,4,7])
Lợi_Nhuận['Cao']=fuzz.trimf(Lợi_Nhuận.universe,[8,10,10])
Đơn_kết_Hợp['Ít']=fuzz.trimf(Đơn_kết_Hợp.universe,[0,0,3])
Đơn_kết_Hợp['Một số']=fuzz.trimf(Đơn_kết_Hợp.universe,[2,4,7])
Đơn_kết_Hợp['Nhiều']=fuzz.trimf(Đơn_kết_Hợp.universe,[8,10,10])
Ưu_Tiên['Thấp']=fuzz.trimf(Ưu_Tiên.universe,[0,0,3])
Ưu_Tiên['Trung Bình']=fuzz.trimf(Ưu_Tiên.universe,[2,4,7])
Ưu_Tiên['Cao']=fuzz.trimf(Ưu_Tiên.universe,[8,10,10])

Mật_Độ_Đơn.view()
Delivery_Urgency.view()
Tải_Trọng.view()
Traffic_Condition.view()
Lợi_Nhuận.view()
Đơn_kết_Hợp.view()
```
